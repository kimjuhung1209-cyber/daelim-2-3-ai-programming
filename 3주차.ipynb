반복문과 조건문 함께 사용

```python
scores = [42, 60 , 77, 89, 20] 

for score in scores:
    if score >= 60:
        print(score, "합격")
    else:
        print(score, "불합격")
```

```python
scores = [65, 82, 91, 100, 58, 77]

#70점 이상은 합격
#출력: 합격자는 x명 입니다.

count = 0

for a in scores:
    if a >= 70:
        count += 1

print(f'합격자는 {count}명 입니다')
```
